```python
diabetes['Pregnancies']=diabetes['Pregnancies'].fillna(diabetes['Pregnancies'].median())
diabetes['Glucose']=diabetes['Glucose'].fillna(diabetes['Glucose'].median())
diabetes['BloodPressure']=diabetes['BloodPressure'].fillna(diabetes['BloodPressure'].median())
diabetes['SkinThickness']=diabetes['SkinThickness'].fillna(diabetes['SkinThickness'].median())
diabetes['Insulin']=diabetes['Insulin'].fillna(diabetes['Insulin'].median())
diabetes['BMI']=diabetes['BMI'].fillna(diabetes['BMI'].median())
diabetes['DiabetesPedigreeFunction']=diabetes['DiabetesPedigreeFunction'].fillna(diabetes['DiabetesPedigreeFunction'].median())
diabetes['Age']=diabetes['Age'].fillna(diabetes['Age'].median())
```

```python
x_diabetes = diabetes.drop('Outcome', axis=1)
y_diabetes = diabetes['Outcome']
print("input shape:",x_diabetes.shape)
print("target shape:",y_diabetes.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x_diabetes,y_diabetes,test_size=0.2,random_state=42)
print("train input shape:",x_train.shape)
print("train target shape:",y_train.shape)
print("test input shape:",x_test.shape)
print("test target shape:",y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
decision_tree = DecisionTreeClassifier(
    max_depth =4,
    class_weight="balanced",
    random_state=42
)
decision_tree.fit(x_train_scaled,y_train)
```

```python
#predict test data
diabetes_predictions = decision_tree.predict(
    x_test_scaled
)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

accuracy = accuracy_score(
    y_test,
    diabetes_predictions
)
print("Accuracy:", accuracy)
print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        diabetes_predictions
    )
)
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        diabetes_predictions
    )
)
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(18,8))
plot_tree(
    decision_tree,
    filled=True,
    feature_names=x_diabetes.columns,
    class_names=["0","1"]
)
plt.title("Decision Tree for diabetes prediction")
plt.show()
```



# **K-Means Clustering – Step-by-Step Solution**

## **1. Setup**

- **K = 2** (Two clusters)
- **Initial Centroid C1 = (1, 1)** (Subject 1)
- **Initial Centroid C2 = (5, 7)** (Subject 4)

**Euclidean Distance Formula:**

\\[ d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2} \\]

## **2. Iteration 1 – Distance to Initial Centroids**

| Subject | Point (A,B) | Distance to C1 | Distance to C2 | Assigned Cluster |
| ------- | ----------- | -------------- | -------------- | ---------------- |
| 1       | (1,1)       | 0.00           | 7.21           | Cluster 1        |
| 2       | (1.5,2)     | 1.12           | 6.10           | Cluster 1        |
| 3       | (3,4)       | 3.61           | 3.61           | Cluster 1\*      |
| 4       | (5,7)       | 7.21           | 0.00           | Cluster 2        |
| 5       | (3.5,5)     | 4.72           | 2.50           | Cluster 2        |
| 6       | (4.5,5)     | 5.32           | 2.06           | Cluster 2        |
| 7       | (3.5,4.5)   | 4.30           | 2.92           | Cluster 2        |

*Subject 3 is equidistant from both centroids, so it is assigned to Cluster 1.*

### **Update Centroids**

**New C1 (Subjects 1, 2, 3):**

\\[ C_1=\left(\frac{1+1.5+3}{3},\frac{1+2+4}{3}\right) \\]

\\[ \boxed{C_1=(1.83,2.33)} \\]

**New C2 (Subjects 4, 5, 6, 7):**

\\[ C_2=\left(\frac{5+3.5+4.5+3.5}{4}, \frac{7+5+5+4.5}{4}\right) \\]

\\[ \boxed{C_2=(4.13,5.38)} \\]

## **3. Iteration 2 – Distance to Updated Centroids**

| Subject | Point (A,B) | Distance to C1 | Distance to C2 | Assigned Cluster |
| ------- | ----------- | -------------- | -------------- | ---------------- |
| 1       | (1,1)       | 1.57           | 5.38           | Cluster 1        |
| 2       | (1.5,2)     | 0.47           | 4.28           | Cluster 1        |
| 3       | (3,4)       | 2.04           | 1.78           | Cluster 2        |
| 4       | (5,7)       | 5.64           | 1.84           | Cluster 2        |
| 5       | (3.5,5)     | 3.15           | 0.73           | Cluster 2        |
| 6       | (4.5,5)     | 3.78           | 0.53           | Cluster 2        |
| 7       | (3.5,4.5)   | 2.74           | 1.08           | Cluster 2        |

### **Update Centroids**

**New C1 (Subjects 1, 2):**

\\[ C_1=\left(\frac{1+1.5}{2},\frac{1+2}{2}\right) \\]

\\[ \boxed{C_1=(1.25,1.50)} \\]

**New C2 (Subjects 3, 4, 5, 6, 7):**

\\[ C_2=\left(\frac{3+5+3.5+4.5+3.5}{5}, \frac{4+7+5+5+4.5}{5}\right) \\]

\\[ \boxed{C_2=(3.90,5.10)} \\]

## **4. Iteration 3 – Final Check (Convergence)**

| Subject | Point (A,B) | Distance to C1 | Distance to C2 | Assigned Cluster |
| ------- | ----------- | -------------- | -------------- | ---------------- |
| 1       | (1,1)       | 0.56           | 5.02           | Cluster 1        |
| 2       | (1.5,2)     | 0.56           | 3.92           | Cluster 1        |
| 3       | (3,4)       | 3.05           | 1.42           | Cluster 2        |
| 4       | (5,7)       | 6.66           | 2.20           | Cluster 2        |

| 6       | (4.5,5)     | 4.78           | 0.61           | Cluster 2        |
| 7       | (3.5,4.5)   | 3.75           | 0.72           | Cluster 2        |

Since no data points changed their clusters, the algorithm stops.

## **5. Final Clusters**

**Cluster 1:**

- Subjects: 1, 2
